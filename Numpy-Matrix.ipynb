```python
import numpy as np
```

```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], 
              [7, 8]])
```

```python
#Matrix Addition
print(A + B)
```

```python
#Matrix Subtraction 
print(A - B)
```

```python
# Element-wise Matrix multiplication
print(A * B)
```

```python
#The @ performs proper matrix multiplication where you multiply rows of the first matrix by columns of the second matrix
result = A @ B 
print(result)
```

```python
print(A.T)
```

```python
#Multiplication using dot product
result_dot = np.dot(A, B)
print(result_dot)
```

```python
from numpy.linalg import det, inv
determinant_A = det(A)
print(determinant_A)
```

```python
determinant_B = det(B)
print(determinant_B)
```

```python
inverse_A = inv(A)
inverse_B = inv(B)
```

```python
print(inverse_A)
print(inverse_B)
```
